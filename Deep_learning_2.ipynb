```python
import numpy as np
import os
import matplotlib.pyplot as plt
import os, random
import pillow_heif
from PIL import Image, ImageChops, ImageEnhance, ImageFilter
import matplotlib.pyplot as plt
```

```python
def show_image(input_dir, title=""):
    file_list = [f for f in os.listdir(input_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))][:20]
    plt.figure(figsize=(20, 5))
    for i, file_name in enumerate(file_list):
        img_path = os.path.join(input_dir, file_name)
        with Image.open(img_path) as img:
            plt.subplot(2, 10, i + 1)
            plt.imshow(img)
            plt.axis('off')

    plt.suptitle(title, fontsize=16)
    plt.tight_layout()
    plt.show()
```

```python
day_Resized = r'D:\KULIAH\SEMESTER 5\DEEP LEARNING\PROJECT\DATASET\DATA SEKUNDER\Resized\Day'
night_Resized = r'D:\KULIAH\SEMESTER 5\DEEP LEARNING\PROJECT\DATASET\DATA SEKUNDER\Resized\Night'

os.makedirs(day_Resized, exist_ok=True)
os.makedirs(night_Resized, exist_ok=True)

def resize_images(input_dir, output_dir, size=(256, 256)):
    for filename in os.listdir(input_dir):
        if filename.lower().endswith(('.jpg', '.jpeg', '.png')):
            img_path = os.path.join(input_dir, filename)
            try:
                img = Image.open(img_path)
                img_resized = img.resize(size)
                save_path = os.path.join(output_dir, filename)
                img_resized.save(save_path)
            except Exception as e:
                print(f"Gagal memproses {filename}: {e}")

resize_images(day, day_Resized)
resize_images(night, night_Resized)
```

```python
def visualize_RGB_distribution(image_path, title):
    img = np.array(Image.open(image_path).convert('RGB'))
    img = (img / 255.0)

    colors = ['red', 'green', 'blue']
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 4, 1)
    plt.imshow(img)
    plt.title(title)
    plt.axis("off")

    for i, color in enumerate(colors):
        plt.subplot(1, 4, i + 2)
        plt.hist((img[:, :, i] * 255).flatten(), bins=50, color=color, alpha=0.7)
        plt.title(f"{color.capitalize()} Channel")

    plt.tight_layout()
    plt.show()

day_sample = os.path.join(day_Resized, os.listdir(day_Resized)[0])
night_sample = os.path.join(night_Resized, os.listdir(night_Resized)[0])
```

```python
def get_image_paths(input_dir):
    return [os.path.join(input_dir, f) for f in os.listdir(input_dir)
            if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

def load_image(path):
    with Image.open(path) as img:
        return np.array(img.convert("RGB"))

day_images = get_image_paths(day_Resized)
night_images = get_image_paths(night_Resized)
```

```python
def get_image_sizes(image_paths):
    widths, heights = [], []
    for path in image_paths:
        with Image.open(path) as img:
            w, h = img.size
            widths.append(w)
            heights.append(h)
    return np.array(widths), np.array(heights)

day_w, day_h = get_image_sizes(day_images)
night_w, night_h = get_image_sizes(night_images)

plt.figure(figsize=(8, 6))
plt.scatter(day_w, day_h, alpha=0.5, color='skyblue', label='Day')
plt.scatter(night_w, night_h, alpha=0.5, color='orange', label='Night')
plt.xlabel('Width (px)')
plt.ylabel('Height (px)')
plt.title('Distribusi Resolusi Gambar (Day vs Night)')
plt.legend()
plt.show()
```

```python
def compute_brightness(image_paths, sample_size=200):
    brightness = []
    for path in image_paths[:sample_size]:
        img = load_image(path)
        brightness.append(np.mean(img))
    return np.array(brightness)

day_brightness = compute_brightness(day_images)
night_brightness = compute_brightness(night_images)

plt.figure(figsize=(8, 5))
plt.hist(day_brightness, bins=30, alpha=0.6, color='skyblue', label='Day')
plt.hist(night_brightness, bins=30, alpha=0.6, color='orange', label='Night')
plt.axvline(np.mean(day_brightness), color='blue', linestyle='--', label='Mean Day')
plt.axvline(np.mean(night_brightness), color='red', linestyle='--', label='Mean Night')
plt.title('Distribusi Brightness & Potensi Outlier')
plt.xlabel('Brightness (mean pixel intensity)')
plt.ylabel('Frekuensi')
plt.legend()
plt.show()
```

```python
def preview_images(input_dir, title="", jumlah=20):
    file_list = [f for f in os.listdir(input_dir)
                 if f.lower().endswith(('.jpg', '.jpeg', '.png'))][:jumlah]
    cols = 10
    rows = (len(file_list) + cols - 1) // cols
    plt.figure(figsize=(20, 2.5 * rows))
    for i, file_name in enumerate(file_list):
        img_path = os.path.join(input_dir, file_name)
        with Image.open(img_path) as img:
            plt.subplot(rows, cols, i + 1)
            plt.imshow(img)
            plt.axis('off')
    plt.suptitle(title, fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()

preview_images(day_Resized, title="Contoh Gambar Day")
preview_images(night_Resized, title="Contoh Gambar Night")
```
